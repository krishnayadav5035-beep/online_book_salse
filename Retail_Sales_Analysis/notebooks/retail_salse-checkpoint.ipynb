```python
print("Total Orders:", len(orders))
print("Total Customers:", len(customers))
print("Total Products:", len(products))

print("Total Revenue:", orders["payment_amount"].sum())

print("Average Order Value:", orders["payment_amount"].mean())
```

```python
# Convert order_date into datetime
orders["order_date"] = pd.to_datetime(orders["order_date"])

# Calculate monthly revenue
monthly_revenue = (
    orders.groupby(orders["order_date"].dt.to_period("M"))["payment_amount"]
    .sum()
)

# Create monthly analysis table
monthly_analysis = monthly_revenue.to_frame(name="Revenue")

# Calculate Month-on-Month growth
monthly_analysis["MoM_Growth_%"] = (
    monthly_analysis["Revenue"].pct_change() * 100
)

print(monthly_analysis.round(2))

```

```python

import matplotlib.pyplot as plt

print(plt)
print(hasattr(plt, "figure"))

plt.figure(figsize=(14, 5))

plt.plot(
    monthly_analysis.index.astype(str),
    monthly_analysis["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()

plt.savefig("monthly_revenue.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
product_sales = orders.merge(
    products,
    on="product_id",
    how="left"
)
```

```python
top_products = (
    product_sales.groupby("book_title")["quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
top_products.plot(kind="bar", figsize=(12, 5))

plt.title("Top 10 Best-Selling Products")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
plt.savefig("top_products.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
category_revenue = (
    product_sales.groupby("genre")["payment_amount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_revenue)
```

```python
plt.figure(figsize=(10, 5))

category_revenue.plot(kind="bar")

plt.title("Category-wise Revenue")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show
plt.savefig("category_revenue.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
category_share = (
    category_revenue / category_revenue.sum()
) * 100

print(category_share.round(2))
```

```python
plt.figure(figsize=(10, 5))

category_share.plot(kind="bar")

plt.title("Share of total revenue by product genre")
plt.xlabel("Category")
plt.ylabel("Total persentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
